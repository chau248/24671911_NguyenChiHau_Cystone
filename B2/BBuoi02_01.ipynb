## Phần 2: Lọc và sắp xếp dữ liệu

**Câu 6.** Lọc SALES > 5000, đếm số đơn.

```python
df[df['SALES'] > 5000]
```

```python
len(df[df['SALES'] > 5000])
```

**Câu 7.** Lọc USA + DEALSIZE Large, sắp xếp SALES giảm dần, lấy 10 dòng đầu.

```python
df[(df['COUNTRY'] == 'USA') & (df['DEALSIZE'] == 'Large')].sort_values('SALES', ascending=False).head(10)
```

**Câu 8.** Lọc Profit âm, value_counts STATUS và DEALSIZE.

```python
df_profit = df[df['Profit'] < 0]
print(df_profit['STATUS'].value_counts())
print(df_profit['DEALSIZE'].value_counts())
```

**Nhận xét:** Lỗ tập trung nhiều nhất ở `STATUS = Shipped` (454 đơn) và `DEALSIZE = Medium/Small`, không phải do huỷ/tranh chấp.

**Câu 9.** Lọc Discount >= 0.3, so sánh SALES trung bình với toàn bộ.

```python
df[df['Discount'] >= 0.3]['SALES'].mean()
```

## Phần 3: Tạo cột mới và biến đổi dữ liệu

**Câu 10.** Tạo `ORDER_YEAR`, `ORDER_MONTH`, `ORDER_QUARTER` từ `ORDERDATE`.

```python
df['ORDER_YEAR'] = df['ORDERDATE'].dt.year
df['ORDER_MONTH'] = df['ORDERDATE'].dt.month
df['ORDER_QUARTER'] = df['ORDERDATE'].dt.quarter
df.head()
```

**Câu 11.** Tạo `PROFIT_MARGIN`, tìm đơn cao nhất/thấp nhất.

```python
df['PROFIT_MARGIN'] = df['Profit'] / df['SALES']
```

```python
df.sort_values('PROFIT_MARGIN', ascending=False).head(1)
```

```python
df.sort_values('PROFIT_MARGIN').head(1)
```

**Câu 12.** Tạo `NET_SALES`, so sánh tổng SALES và tổng NET_SALES.

```python
df['NET_SALES'] = df['SALES'] * (1 - df['Discount'])
```

```python
df['SALES'].sum(), df['NET_SALES'].sum()
```

## Phần 4: Nhóm dữ liệu (Groupby) và tổng hợp

**Câu 13.** Tổng SALES và Profit theo COUNTRY, Top 5 doanh số.

```python
df.groupby(by='COUNTRY')['SALES'].sum()
```

```python
df.groupby(by='COUNTRY')[['SALES', 'Profit']].sum()
```

```python
df.groupby('COUNTRY')['SALES'].sum().sort_values(ascending=False).head(5)
```

**Câu 14.** SALES và Profit trung bình theo DEALSIZE.

```python
df.groupby('DEALSIZE')[['SALES', 'Profit']].mean()
```

**Nhận xét:** Large có SALES trung bình cao nhất nhưng Profit trung bình lại thấp nhất; Medium mới là nhóm lãi trung bình cao nhất.

**Câu 15.** Số đơn duy nhất theo STATUS, tỷ lệ % Cancelled/Disputed.

```python
df.groupby('STATUS')['ORDERNUMBER'].nunique()
```

```python
df.groupby('STATUS')['ORDERNUMBER'].nunique()[['Cancelled', 'Disputed']].sum() / df['ORDERNUMBER'].nunique() * 100
```

**Câu 16.** `pivot_table` tổng SALES theo COUNTRY và DEALSIZE.

```python
pd.pivot_table(df, values='SALES', index='COUNTRY', columns='DEALSIZE', aggfunc='sum')
```

**Câu 17.** Tổng SALES theo ORDER_YEAR, năm nào cao nhất.

```python
df.groupby('ORDER_YEAR')['SALES'].sum()
```

```python
df.groupby('ORDER_YEAR')['ORDER_MONTH'].nunique()
```

**Nhận xét:** Doanh số cao nhất là năm 2004. Năm 2005 chỉ có dữ liệu đến tháng 5 (5/12 tháng) nên không so sánh trực tiếp với 2003/2004 được.
